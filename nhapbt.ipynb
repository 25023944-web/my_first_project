```python
n = int(input("Nhập chiều cao tam giác: "))

# Vẽ đỉnh tam giác
if n > 0:
    print(" " * (n - 1) + "*")

# Vẽ các hàng ở giữa (hollow part)
for i in range(1, n - 1):
    khoang_cach_ngoai = " " * (n - i - 1)
    khoang_cach_trong = " " * (2 * i - 1)
    print(f"{khoang_cach_ngoai}*{khoang_cach_trong}*")

# Vẽ đáy tam giác
if n > 1:
    print("*" * (2 * n - 1))
```

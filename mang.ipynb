nhập mảng 1 chiều

sắp xếp nhanh

```python
n=int(input())
a=list(map(int,input().split()))
def sxn(T,P):
    global a
    i=T
    j=P
    x=a[(T+P)//2]
    while (i<=j):
        while a[i]>x:i+=1
        while a[j]<x:j-=1
        if i<=j:
            a[i],a[j]=a[j],a[i]
            i+=1
            j-=1
    if T<j:sxn(T,j)
    if i<P:sxn(i,P)
sxn(0,n-1)
print(a)
```
